```python
import polars as pl
import matplotlib.pyplot as plt
# Step 1: Read column names from line 2
with open("TOA5_23861_Time_Series_672_2022_12_15_0000.dat", "r") as f:
    lines = f.readlines()
    column_names = [col.strip().strip('"') for col in lines[1].split(",")]

# Step 2: Read data, skipping the first 4 metadata rows
df = pl.read_csv(
    "TOA5_23861_Time_Series_672_2022_12_15_0000.dat",
    skip_rows=4,
    has_header=False,              # Since we manually provide column names
    new_columns=column_names,
    quote_char='"',
)

def normalize_ts_column(series: pl.Series) -> pl.Series:
    def pad_microseconds(ts: str) -> str:
        if '.' not in ts:
            return ts + '.000000'
        base, _, frac = ts.partition('.')
        return base + '.' + frac.ljust(6, '0')
    return pl.Series(series.name, [pad_microseconds(t) for t in series])

cleaned_ts = normalize_ts_column(df["TIMESTAMP"])
df = df.with_columns(
    cleaned_ts.str.strptime(pl.Datetime, format="%Y-%m-%d %H:%M:%S%.f", strict=False).alias("TIMESTAMP")
)
```

```python
# Step 3: Extract 'Uz' for plotting
uz = df["Uz"].to_numpy()
tm = df["TIMESTAMP"].to_numpy()

# Step 4: Plot
plt.figure(figsize=(12, 4))
plt.plot(tm, uz, linewidth=0.5, label="Uz")
plt.title("Vertical Wind Component (Uz)")
plt.xlabel("Sample Index")
plt.ylabel("Uz [m/s]")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import polars as pl

def rolling_filter(df, col, ts='TIMESTAMP')

# Step 1: Ensure TIMESTAMP is sorted and typed correctly
df = df.sort("TIMESTAMP")
df = df.with_columns(
    pl.col("TIMESTAMP").cast(pl.Datetime)
)

# Step 2: Calculate rolling mean and std using a 5s time window
roll = (
    df.rolling(index_column="TIMESTAMP", period="5s", closed="both")
    .agg([
        pl.col("Uz").mean().alias("Uz_roll_mean"),
        pl.col("Uz").std().alias("Uz_roll_std")
    ])
)

# Step 3: Join rolling stats back to the original DataFrame
df = df.join(roll, on="TIMESTAMP", how="left")

# Step 4: Null out any spike that is more than ±3σ from the rolling mean
df = df.with_columns([
    pl.when(
        (pl.col("Uz") < pl.col("Uz_roll_mean") - 3 * pl.col("Uz_roll_std")) |
        (pl.col("Uz") > pl.col("Uz_roll_mean") + 3 * pl.col("Uz_roll_std"))
    )
    .then(None)
    .otherwise(pl.col("Uz"))
    .alias("Uz_filtered")
])
```

```python
# Step 3: Extract 'Uz' for plotting
uz = df["Uz_filtered"].to_numpy()
tm = df["TIMESTAMP"].to_numpy()

# Step 4: Plot
plt.figure(figsize=(12, 4))
plt.plot(tm, uz, linewidth=0.5, label="Uz")
plt.title("Vertical Wind Component (Uz)")
plt.xlabel("Sample Index")
plt.ylabel("Uz [m/s]")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import polars as pl
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import statsmodels.api as sm

# Assume you already have your filtered DataFrame called df
# with column "Uz"

# Convert to numpy array
uz = df["Uz"].to_numpy()

# --- Shapiro-Wilk test ---
shapiro_stat, shapiro_p = stats.shapiro(uz)
print(f"Shapiro-Wilk test: W={shapiro_stat:.3f}, p={shapiro_p:.3e}")

# --- Anderson-Darling test ---
anderson_result = stats.anderson(uz, dist='norm')
print("Anderson-Darling test statistic:", anderson_result.statistic)
print("Critical values:", anderson_result.critical_values)
print("Significance levels:", anderson_result.significance_level)

# --- Kolmogorov-Smirnov test ---
ks_stat, ks_p = stats.kstest(
    (uz - np.mean(uz)) / np.std(uz, ddof=1), "norm"
)
print(f"Kolmogorov-Smirnov test: D={ks_stat:.3f}, p={ks_p:.3e}")

# --- Q-Q Plot ---
sm.qqplot(uz, line='45')
plt.title("Q-Q Plot of Uz")
plt.show()
```

```python
import polars as pl
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import statsmodels.api as sm

def normality_check(
    df: pl.DataFrame,
    col: str,
    sample_for_shapiro: int = 5000,   # Shapiro is most reliable <= ~5000
    qqplot: bool = True,
    random_seed: int = 0
):
    """
    Run normality diagnostics on a Polars column.
    Returns a dict of results and (optionally) shows a Q–Q plot.
    """

    # 1) Pull clean values from Polars (drop nulls/NaNs)
    s = df.select(pl.col(col).drop_nans().drop_nulls()).to_series()
    x = s.to_numpy()
    n = x.size
    if n == 0:
        raise ValueError(f"Column '{col}' has no non-null values.")

    # 2) Basic stats in Polars (mean/std/skew/kurtosis)
    basic = df.select([
        pl.col(col).mean().alias("mean"),
        pl.col(col).std().alias("std"),
        pl.col(col).skew().alias("skewness"),
        pl.col(col).kurtosis().alias("kurtosis"),  # Fisher’s definition → normal ≈ 0
    ]).row(0)

    mean, std, skewness, kurtosis = basic

    # 3) Shapiro–Wilk (subsample if large)
    rng = np.random.default_rng(random_seed)
    if n > sample_for_shapiro:
        x_shapiro = rng.choice(x, size=sample_for_shapiro, replace=False)
    else:
        x_shapiro = x
    shapiro_W, shapiro_p = stats.shapiro(x_shapiro)

    # 4) Anderson–Darling (for normal)
    ad = stats.anderson(x, dist='norm')  # ad.statistic, ad.critical_values, ad.significance_level

    # 5) Kolmogorov–Smirnov vs. fitted normal (normalize first)
    xz = (x - mean) / (std if std > 0 else 1.0)
    ks_D, ks_p = stats.kstest(xz, "norm")

    # 6) Optional Q–Q plot
    if qqplot:
        sm.qqplot(x, line='45')
        plt.title(f"Q–Q Plot of {col}")
        plt.xlabel("Theoretical quantiles")
        plt.ylabel("Sample quantiles")
        plt.show()

    # 7) Pretty print + return results
    print(f"\nNormality diagnostics for '{col}' (n={n:,}):")
    print(f"  mean={mean:.4g}, std={std:.4g}, skewness={skewness:.4g}, kurtosis(Fisher)={kurtosis:.4g}")
    print(f"  Shapiro–Wilk: W={shapiro_W:.4f}, p={shapiro_p:.3e}  (sampled n={len(x_shapiro):,})")
    print(f"  Anderson–Darling: A2={ad.statistic:.4f}")
    for lvl, crit in zip(ad.significance_level, ad.critical_values):
        print(f"    critical @ {lvl:.1f}%: {crit:.4f}  → {'ACCEPT normal' if ad.statistic < crit else 'REJECT normal'}")
    print(f"  Kolmogorov–Smirnov vs N(0,1): D={ks_D:.4f}, p={ks_p:.3e}")

    return {
        "n": n,
        "mean": float(mean),
        "std": float(std),
        "skewness": float(skewness),
        "kurtosis_fisher": float(kurtosis),
        "shapiro": {"W": float(shapiro_W), "p": float(shapiro_p), "n_tested": int(len(x_shapiro))},
        "anderson_darling": {
            "statistic": float(ad.statistic),
            "critical_values": ad.critical_values.tolist(),
            "significance_levels": ad.significance_level.tolist(),
        },
        "kolmogorov_smirnov": {"D": float(ks_D), "p": float(ks_p)},
    }
```

```python
# After your ±3σ filtering step
res = normality_check(df, "Uz", sample_for_shapiro=5000, qqplot=True)

# Example quick checks:
print("Skew near 0?", abs(res["skewness"]) < 0.2)
print("Kurtosis near 0?", abs(res["kurtosis_fisher"]) < 0.5)
```

```python
import polars as pl
from datetime import timedelta

def save_three_hour_sample_subset(
    df: pl.DataFrame,
    timestamp_col: str = "TIMESTAMP",
    columns_to_keep: list[str] = None,
    basename: str = "sample"
) -> tuple[str, str]:
    """
    Extracts a 3-hour slice and selected columns from a Polars DataFrame, then saves as CSV and Parquet.

    Parameters
    ----------
    df : pl.DataFrame
        The input Polars DataFrame.
    timestamp_col : str
        Name of the datetime column (default: "TIMESTAMP").
    columns_to_keep : list of str
        Subset of columns to retain (default: all).
    basename : str
        Base name for the output files (without extension).

    Returns
    -------
    tuple[str, str]
        Paths to the saved CSV and Parquet files.
    """
    # Cast timestamp to Datetime and sort
    df = df.with_columns(pl.col(timestamp_col).cast(pl.Datetime))
    df = df.sort(timestamp_col)

    # Compute 3-hour window
    start_time = df[timestamp_col].min()
    end_time = start_time + timedelta(hours=3)

    # Filter for 3-hour slice
    df_sample = df.filter(
        (pl.col(timestamp_col) >= start_time) & 
        (pl.col(timestamp_col) < end_time)
    )

    # Keep only selected columns
    if columns_to_keep:
        df_sample = df_sample.select(columns_to_keep)

    # Save to disk
    csv_path = f"{basename}.csv"
    parquet_path = f"{basename}.parquet"

    df_sample.write_csv(csv_path)
    df_sample.write_parquet(parquet_path)

    print(f"✅ Saved 3-hour subset sample:")
    print(f"  • CSV:     {csv_path}")
    print(f"  • Parquet: {parquet_path}")
    
    return csv_path, parquet_path
```
